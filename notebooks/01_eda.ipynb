Exploratory data analysis

basic cleaning

```python
df = df.drop(columns="id")
df = df[df["gender"] != "Other"].reset_index(drop=True)
print(df.shape)
```

```python
cat_cols = ["gender","ever_married", "work_type", "Residence_type", "smoking_status"]
fig, axes = plt.subplots(2,3,figsize=(15, 8))
axes = axes.ravel()
for ax, col in zip(axes, cat_cols):
    rate = df.groupby(col)["stroke"].mean().mul(100).sort_values()
    sns.barplot(x=rate.values, y=rate.index, ax=ax, color="steelblue")
    ax.set_title(f"Stroke rate (%) by {col}")
    ax.set_xlabel("Stroke rate (%)")
    ax.set_ylabel("")
axes[-1].axis("off")
plt.tight_layout()
plt.savefig("../figures/stroke_rate_categorical.png", dpi=150)
plt.show()
```

```python
numeric distributions
```

```python
num_cols = ["age", "avg_glucose_level", "bmi"]
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, num_cols):
    sns.kdeplot(data=df, x=col, hue="stroke", common_norm=False, fill=True, ax=ax)
    ax.set_title(f"{col} by stroke")
    plt.tight_layout()
plt.savefig("../figures/numeric_distributions.png", dpi=150)
plt.show()
```

stroke rate by age group

```python
age_bins = [0, 18, 40, 50, 60, 70, 80, 90]
rate_by_age = df.groupby(pd.cut(df["age"], bins=age_bins, right=False), observed=True)["stroke"].mean().mul(100)
rate_by_age.plot(kind="bar", color="steelblue", figsize=(8, 4))
plt.ylabel("Stroke rate (%)")
plt.title("Stroke rate by age group")
plt.tight_layout()
plt.savefig("../figures/stroke_rate_by_age.png", dpi=150)
plt.show()
```

hypertension and heart disease

```python
for col in ["hypertension", "heart_disease"]:
    print(df.groupby(col)["stroke"].agg(["mean", "sum", "count"]).round(3), "\n")
```

correlations

```python
corr_cols = ["age", "hypertension", "heart_disease", "avg_glucose_level", "bmi", "stroke"]
plt.figure(figsize=(7, 5))
sns.heatmap(df[corr_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation matrix")
plt.tight_layout()
plt.savefig("../figures/correlation_matrix.png", dpi=150)
plt.show()
```
